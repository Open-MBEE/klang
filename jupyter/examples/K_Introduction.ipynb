# K Language - Interactive Constraint Programming

This notebook demonstrates the K constraint programming language.
K models are solved using the Z3 SMT solver.

```k
%help
```

## Example 1: Simple Constraints

```k
x : Int
y : Int

req x + y = 10
req x < y
req x > 0
```

```k
class Point {
    x : Int
    y : Int
}

p : Point
req p.x + p.y = 100
```
