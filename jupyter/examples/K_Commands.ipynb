# K Language - Complete Command Reference

This notebook demonstrates all magic commands and K language features.

## Magic Commands

### %help - Show help

```k
%help
```

### %reset - Clear the model

```k
%reset
```

### Basic Example - Int constraints

```k
x : Int
y : Int
req x + y = 10
req x < y
```

### %show - Display accumulated code

```k
%show
```

### %verbose - Toggle verbose output

```k
%verbose on
```

```k
%reset
z : Int
req z = 42
```

```k
%verbose off
```

### %stats - Show statistics

```k
%stats
```

### %raw - Show raw K output

```k
%raw
```

### %timeout - Set solver timeout

```k
%timeout 60
```

---
## K Language Types

### Int - Integers

```k
%reset
n : Int
req n > 0
req n < 100
req n % 7 = 0
```

### Real - Real numbers

```k
%reset
r : Real
req r > 0.0
req r < 1.0
```

### Bool - Booleans

```k
%reset
a : Bool
b : Bool
req a or b
req not (a and b)
```

### String - Strings

```k
%reset
s : String
req s.length() > 0
req s.length() < 10
```

### Time - Time values

```k
%reset
t : Time
req t > 0
```

### Duration - Duration values

```k
%reset
d : Duration
req d > 0
```

### Classes

```k
%reset
class Point {
    x : Int
    y : Int
    req x >= 0
    req y >= 0
}

p : Point
req p.x + p.y = 100
```

### UNSAT Example

```k
%reset
x : Int
req x > 10
req x < 5
```
